```python
# Load the Moneyball dataset
file_path = "content/drive/MyDrive/Workshop/Lab/module1/lab1/data/baseball.csv"
moneyball_df = pd.read_csv(file_path)

# Display the first few rows and the structure of the dataset
print(moneyball_df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Filter dataset to drop missing values (if needed)
df = moneyball_df.dropna(subset=['OBP', 'SLG', 'W'])

# Step 1: Correlation Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df[['W', 'OBP', 'SLG', 'BA', 'RS', 'RA']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation of Team Stats with Wins')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot of OBP vs Wins
plt.figure(figsize=(10, 6))
sns.regplot(data=moneyball_df, x='OBP', y='W', scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Relationship between On-Base Percentage (OBP) and Wins', fontsize=20)
plt.xlabel('On-Base Percentage (OBP)', fontsize=20)
plt.ylabel('Wins', fontsize=20)
plt.grid(True)
plt.tight_layout()
plt.show()
```
